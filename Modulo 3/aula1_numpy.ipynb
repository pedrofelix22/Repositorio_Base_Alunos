# **Trabalando com Numpy**


```python
#como criar um array Numpay?

array_notas = np.array([7,8,9,10])


print(array_notas + 10 / 4 + 3 / 2 )
```

```python
print(array_notas)
print(array_notas > 8 )
#filtragem de -> Mascara de Valores
print(array_notas[array_notas > 8])
```
